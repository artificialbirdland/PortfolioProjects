# Enhancing Accessibility and Uniqueness of Hashtags in Instagram Dataset

```python
# Importing the libraries
import pandas as pd  # Import the pandas library, which is used for data manipulation and analysis

# Installing the csv file
file_path = 'D:\\file.csv'  # Specify the file path of the CSV file

df = pd.read_csv(file_path, encoding='latin1')  # Read the CSV file into a DataFrame using the specified encoding
```

## Adding Unique Identifiers to DataFrame Rows

```python
# Inserting a new column to uniquely identify each row

df.insert(0, 'ID', range(1, 1 + len(df)))  # Insert a new column at the beginning of the DataFrame with unique IDs for each row

pd.set_option('display.max_columns', None)  # Set the option to display all columns when printing the DataFrame
pd.set_option('display.max_colwidth', None)  # Set the option to display the full content of each column without truncation

df.head()  # Display the first few rows of the DataFrame
```

## Splitting Hashtags and Creating a New DataFrame with Unique Identifiers

```python
new_df = []  # Initialize an empty list to store the new DataFrame
new_data = []  # Initialize an empty list to store the processed data

# Process the 'Hashtags' column in each row
for index, row in df.iterrows():  # Iterate over each row in the original DataFrame
    hashtags = row['Hashtags']  # Get the value of the 'Hashtags' column for the current row
    if pd.notna(hashtags):  # Check if the value is not NaN (not empty)
        hashtag_list = hashtags.split()  # Split the hashtags string into a list of individual hashtags
        for hashtag in hashtag_list:  # Iterate over each hashtag in the list
            new_data.append({'Original_Row_ID': index + 1, 'Hashtag': hashtag})  # Append a dictionary with the original row ID and hashtag to new_data

# Create the new DataFrame from the processed data
new_df = pd.DataFrame(new_data)

# Add an 'ID' column to the new DataFrame
new_df.insert(0, 'ID', range(1, 1 + len(new_df)))  # Insert a new column at the beginning with unique IDs for each row

# Display the new DataFrame
new_df.head(50) # Print the first 50 rows of the new DataFrame to verify the results
```

## Cleaning and Resetting DataFrame for Hashtags

```python
new_df = new_df.drop(columns=['ID']).reset_index(drop=True)  # Remove the 'ID' column and reset the index of the DataFrame without adding a new index column
```

```python
# Remove the '#' character from the 'Hashtag' column
new_df['Hashtag'] = new_df['Hashtag'].str.replace('#', '')

# Display the first 50 rows of the updated DataFrame
print(new_df.head(50))
```

## Assigning Unique IDs to Hashtags

```python
# Determine unique hashtags
unique_hashtags = new_df['Hashtag'].unique()  # Get an array of unique hashtags from the 'Hashtag' column

# Assign a 3-digit ID to each unique hashtag
hashtag_to_id = {hashtag: f"{i:03}" for i, hashtag in enumerate(unique_hashtags, 1)}  # Create a dictionary mapping each unique hashtag to a 3-digit ID

# Add a new column to the original DataFrame containing these IDs
new_df['Hashtag_ID'] = new_df['Hashtag'].map(hashtag_to_id)  # Map each hashtag in the 'Hashtag' column to its corresponding ID and add it to a new column

# Display the first 50 rows of the updated DataFrame
new_df.head(50)
```

```python
# Save the changes to a new file (optional)
new_df.to_csv('D:\\Projects', index=False)
```
